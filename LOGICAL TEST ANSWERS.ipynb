```python
#Q1
data = [12, 7, 12, 5, 7, 9, 5, 12, 9, 7, 15]

duplicates = []

for value in data:
    if data.count(value) > 1 and value not in duplicates:
        duplicates.append(value)

print("Duplicate Values:")

for value in duplicates:
    print(value, "->", data.count(value), "times")
```

```python
#Q2
ids = [1, 2, 3, 4, 5, 7, 8, 9, 10]

total = 0

for i in range(1, 11):
    total = total + i

actual_total = 0

for id in ids:
    actual_total = actual_total + id

missing_id = total - actual_total

print("Missing ID:", missing_id)
```
